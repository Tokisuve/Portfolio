```python
xyz = torch.arange(1., 9.)
xyz
```

```python
xyz.shape
```

```python
xyz.reshape(2,4)
```

```python
new_xyz = xyz.view(2,2,2)

print(new_xyz)
print('1. first row from first block: ', new_xyz[0,0,:]) # Same as new_xyz[0,0] or new_xyz[0][0]
print('2. first row from second block: ', new_xyz[1,0,:]) # Same as new_xyz[1,0] or new_xyz[1][0]
print('3. first row from each block: ', new_xyz[:,0])
print('4. first column from first block: ', new_xyz[0,:,0])
print('5. first column from second block: ', new_xyz[1,:,0])
print('6. first column from each block: ', new_xyz[:,:,0])
print('7. first column from each block: ', new_xyz[:,0,0])
```

```python
# If we wanted to stack our new tensor on top of itself five times, we could do so with torch.stack().

stacked_new_xyz = torch.stack([new_xyz, new_xyz, new_xyz, new_xyz, new_xyz], dim=0)
stacked_new_xyz, stacked_new_xyz.shape
```

```python
# permute() is used to reorder the dimensions (axes) of a tensor.
```

```python
# Create tensor with specific shape
x_original = torch.rand(size=(224, 224, 3))

# Permute the original tensor to rearrange the axis order
x_permuted = x_original.permute(2, 0, 1) # shifts axis 0->1, 1->2, 2->0

print(f"Previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")
```
